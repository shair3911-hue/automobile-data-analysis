## Read Data

```python
import pandas as pd
df = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data", header=None)
print(df.head())
df.head()
```

```python
df.tail()
```

## Add Headers

### Step 1- create header list

```python
headers = ["symboling","normalized-losses","make","fuel-type","aspiration",
           "num-of-doors","body-style","drive-wheels","engine-location",
           "wheel-base","length","width","height","curb-weight","engine-type",
           "num-of-cylinders","engine-size","fuel-system","bore","stroke",
           "compression-ratio","horsepower","peak-rpm","city-mpg",
           "highway-mpg","price"]
```

### Step 2- Asign the headers to the data frame

```python
df.columns= headers   # replaces 0,1,2... with actual names
df.columns #lets verify
```

### Step 3- See the first 10 rows

```python
df.head(10)
```

### Step4- Replace ? with NaN

```python
df1=df.replace("?",np.nan)
df1 #it means replace this with this 
# why np ? its becuase its numpy library for all thses 
#np.nan
#np.array()
#np.mean()
#np.zeros()
```

### Step5 - drop rows where price is missing 

```python
df1.head(20) #assigning again df1 to df 
df= df1.dropna(subset=["price"], axis=0)
# Drop rows where price is missing
# subset=["price"] → only check the price column
# axis=0 → remove ROWS
# Memory trick:
# axis=0 = rows (vertical)
# axis=1 = columns (horizontal)
df.head(20)

```
